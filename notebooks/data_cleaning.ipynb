```python
import pandas as pd

# Load the data on referenda
referenda_results_new = pd.read_csv('../data/raw/referenda_results.csv')

# Visualise a summary of the DataFrame
referenda_results_new_info = referenda_results_new.info()
referenda_results_new_head = referenda_results_new.head()
referenda_results_new_info, referenda_results_new_head
```

```python
# Remove columns containing only null values
columns_to_drop = [
    'title_fr_CH', 'title_it_CH', 'title_rm_CH', 'short_title_de_CH', 
    'short_title_fr_CH', 'short_title_it_CH', 'short_title_rm_CH', 'expats'
]
referenda_results_cleaned = referenda_results_new.drop(columns=columns_to_drop)

# Convert 'date' column to datetime format
referenda_results_cleaned['date'] = pd.to_datetime(referenda_results_cleaned['date'])

# Summary of missing values
missing_values_summary = referenda_results_cleaned.isnull().sum()
referenda_results_cleaned_head = referenda_results_cleaned.head()
missing_values_summary, referenda_results_cleaned_head
```

- Columns that were entirely null or not relevant have been dropped.

- The date column has been converted to datetime format.

- There are 126 missing values in the district and name columns.

```python
# Investigate rows with missing 'district' and 'name' values
missing_district_name = referenda_results_cleaned[referenda_results_cleaned['district'].isnull() | referenda_results_cleaned['name'].isnull()]
missing_district_name_info = missing_district_name.info()
missing_district_name_head = missing_district_name.head()
missing_district_name_info, missing_district_name_head
```

- The district and name columns are entirely missing for these 126 entries.

- These rows have entity_id set to 0, which might indicate a general or non-specific entity (e.g., overall results rather than municipality-specific).

```python
# Drop rows with missing 'district' and 'name' values (entity_id == 0)
referenda_results_cleaned = referenda_results_cleaned[referenda_results_cleaned['entity_id'] != 0]

# Verify the result after dropping
remaining_missing_values_summary = referenda_results_cleaned.isnull().sum()
referenda_results_cleaned_info = referenda_results_cleaned.info()

remaining_missing_values_summary, referenda_results_cleaned_info
```

Cleaned Dataset Summary:

- Total Entries: 23,760
- Columns: 18 
- No Missing Values: All columns are now fully populated.

```python
#Filter data for the date range between 2015 and 2024
referenda_results_filtered = referenda_results_cleaned[
    (referenda_results_cleaned['date'] >= '2018-01-01') &
    (referenda_results_cleaned['date'] <= '2024-12-31')
]

# In 2023 the municipalities of Hemberg, Oberhelfenschwil and Neckertal merged to form the new municipality of Neckertal

# Identify the municipalities to merge
municipalities_to_merge = ['Hemberg', 'Oberhelfenschwil', 'Neckertal']

# Separate the data before 2023 for merging
data_before_2023 = referenda_results_filtered[referenda_results_filtered['date'] < '2023-01-01']
data_from_2023_onwards = referenda_results_filtered[referenda_results_filtered['date'] >= '2023-01-01']

# Aggregate results for the municipalities to be merged into Neckertal before 2023
merged_neckertal_before_2023 = data_before_2023[data_before_2023['name'].isin(municipalities_to_merge)]

# Sum the relevant columns
aggregated_neckertal = merged_neckertal_before_2023.groupby(['id', 'title_de_CH', 'date', 'shortcode', 'domain', 'status', 'answer', 'type', 'ballot_answer'])[['yeas', 'nays', 'invalid', 'empty', 'eligible_voters']].sum().reset_index()

# Add the merged name and entity_id for Neckertal
aggregated_neckertal['name'] = 'Neckertal'
aggregated_neckertal['entity_id'] = merged_neckertal_before_2023['entity_id'].mode()[0]
aggregated_neckertal['district'] = merged_neckertal_before_2023['district'].mode()[0]
aggregated_neckertal['counted'] = merged_neckertal_before_2023['counted'].mode()[0]

# Combine the aggregated data with the rest of the data
data_before_2023_cleaned = data_before_2023[~data_before_2023['name'].isin(municipalities_to_merge)]
referenda_results_final = pd.concat([data_before_2023_cleaned, aggregated_neckertal, data_from_2023_onwards])

# Verify the final result
referenda_results_final_info = referenda_results_final.info()
referenda_results_final_head = referenda_results_final.head()
referenda_results_final_info, referenda_results_final_head
```

- The dataset now includes only data from 2018 to 2024.

- The municipalities of Hemberg, Oberhelfenschwil, and Neckertal have been merged into Neckertal for referendums before 2023.

- Total Entries: 6450

- Columns: 18

- All columns are fully populated.

```python
# Validate and correct data types 

# Check data types
data_types_before = referenda_results_final.dtypes

# Define expected data types
expected_dtypes = {
    'id': 'object',
    'title_de_CH': 'object',
    'date': 'datetime64[ns]',
    'shortcode': 'object',
    'domain': 'object',
    'status': 'object',
    'answer': 'object',
    'type': 'object',
    'ballot_answer': 'object',
    'district': 'object',
    'name': 'object',
    'entity_id': 'int64',
    'counted': 'bool',
    'yeas': 'int64',
    'nays': 'int64',
    'invalid': 'int64',
    'empty': 'int64',
    'eligible_voters': 'int64'
}

# Convert columns to expected data types
for column, dtype in expected_dtypes.items():
    referenda_results_final[column] = referenda_results_final[column].astype(dtype)

# Check data types after conversion
data_types_after = referenda_results_final.dtypes
data_types_before, data_types_after
```

The data types of all columns have been validated and corrected as necessary. The columns now have the appropriate data types:

- Object (String) Columns: id, title_de_CH, shortcode, domain, status, answer, type, ballot_answer, district, name

- Datetime Column: date

- Boolean Column: counted

- Numeric Columns: entity_id, yeas, nays, invalid, empty, eligible_voters

```python
# Save the cleaned referenda results data in data/processed
cleaned_referenda_results_path = '../data/processed/cleaned_referenda_results.csv'
referenda_results_final.to_csv(cleaned_referenda_results_path, index=False)

# Confirm the file is saved
import os
os.path.exists(cleaned_referenda_results_path)
```
